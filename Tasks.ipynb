# Task 1

Write a program that asks the user to enter a student's score (0–100) and prints the letter grade:

Score    Grade
91–100     A
81–90      B
71–80      C
61–70      D
51–60      E
0–50       F


```python
score=int(input("Enter your score: "))
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:     
    print("Grade: C")  
elif score>=60:
    print("Grade: D")
elif score>=50:
    print("Grade: E")
else:
    print("Grade: F") 
```

# Task 2


Write a program that asks the user for a positive integer n and, using a loop, goes through all numbers from 1 to n. For each number:

If it's divisible by both 3 and 5, print " 3 5"
Else if divisible by 3, print "3"
Else if divisible by 5, print "5"
Otherwise, print the number
At the end, print:

the sum of all even numbers from 1 to n
how many numbers were divisible by 3

```python
sum=0
d=0
n=int(input("Enter an integer: "))
for i in range(1,n):
    if i%2==0:
        sum+=i
    if i%3==0:
        d+=1
    if i%3==0 and i%5==0:
        print("3 5")
    elif i%3==0:
        print("3")
    elif i%5==0:
        print("5")
    else:
        print(i)
print(f"The sum of all even numbers from 1 to n: {sum}")
print(f"The number of numbers divisible by 3: {d}")

  
```
